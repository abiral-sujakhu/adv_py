```python

# Q4: Access a SOAP web service using Python


from zeep import Client

WSDL_URL = "http://www.dneonline.com/calculator.asmx?WSDL"

def main():
    client = Client(WSDL_URL)

    # List available operations (optional for lab)
    print("Available SOAP operations:")
    for service in client.wsdl.services.values():
        for port in service.ports.values():
            operations = port.binding._operations.keys()
            print(list(operations))

    # Test a SOAP call
    result = client.service.Add(intA=12, intB=8)
    print("Add(12, 8) =", result)

if __name__ == "__main__":
    main()
```

```python
# Q5: Perform arithmetic operations using SOAP API


from zeep import Client

WSDL_URL = "http://www.dneonline.com/calculator.asmx?WSDL"

def main():
    client = Client(WSDL_URL)

    a = 20
    b = 4

    add_res = client.service.Add(intA=a, intB=b)
    sub_res = client.service.Subtract(intA=a, intB=b)
    mul_res = client.service.Multiply(intA=a, intB=b)
    div_res = client.service.Divide(intA=a, intB=b)

    print(f"Numbers: a={a}, b={b}")
    print("Add:", add_res)
    print("Subtract:", sub_res)
    print("Multiply:", mul_res)
    print("Divide:", div_res)

if __name__ == "__main__":
    main()
```

```python
#question6
import requests

API_KEY = "444b0e4e96bd4d531b186eb390aebaf4"
CITY = "InvalidCityName"   # force error
URL = f"https://api.openweathermap.org/data/2.5/weather?q={CITY}&appid={API_KEY}"

def main():
    try:
        response = requests.get(URL, timeout=10)

        if response.status_code == 200:
            print("Success:")
            print(response.json())

        elif response.status_code == 401:
            print("401 Unauthorized – Invalid API key")

        elif response.status_code == 404:
            print("404 Not Found – City not found")

        elif response.status_code == 429:
            print("429 Too Many Requests – Rate limit exceeded")

        elif 500 <= response.status_code < 600:
            print("Server error:", response.status_code)

        else:
            print("Unexpected status:", response.status_code)

    except requests.RequestException as e:
        print("Request failed:", e)

if __name__ == "__main__":
    main()
```

```python
#question9
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

URL = "https://quotes.toscrape.com/"

def main():
    r = requests.get(URL, timeout=10)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "lxml")

    links = []
    for a in soup.find_all("a", href=True):
        full_link = urljoin(URL, a["href"])  # convert relative to absolute
        links.append(full_link)

    print("Total links found:", len(links))
    print("\nLinks:")
    for link in links:
        print(link)

if __name__ == "__main__":
    main()
```

```python
#question10
import requests
import pandas as pd

URL = "https://www.w3schools.com/html/html_tables.asp"

def main():
    headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept-Language": "en-US,en;q=0.9",
    }

    r = requests.get(URL, headers=headers, timeout=15)
    r.raise_for_status()

    # pandas will directly read HTML tables into DataFrames
    tables = pd.read_html(r.text)
    df = tables[0]  # first table on the page

    print("\n✅ Scraped Table Output (W3Schools)\n")
    print(df.to_string(index=False))

if __name__ == "__main__":
    main()
```

```python
#question11
import requests
from bs4 import BeautifulSoup

URL = "https://quotes.toscrape.com/"

def main():
    r = requests.get(URL, timeout=10)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "lxml")

    quotes = soup.select("span.text")       # all quote text
    authors = soup.select("small.author")   # all authors

    print("First 5 quotes with authors:\n")
    for i in range(min(5, len(quotes), len(authors))):
        print(f"{i+1}. {quotes[i].get_text(strip=True)}")
        print(f"   - {authors[i].get_text(strip=True)}\n")

if __name__ == "__main__":
    main()
```

```python
#question12
import requests
from bs4 import BeautifulSoup
import csv

URL = "https://quotes.toscrape.com/"

def main():
    r = requests.get(URL, timeout=10)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "lxml")

    quote_blocks = soup.select("div.quote")

    data = []
    for block in quote_blocks:
        quote_text = block.select_one("span.text").get_text(strip=True)
        author = block.select_one("small.author").get_text(strip=True)
        data.append((quote_text, author))

    # Save to CSV
    filename = "quotes.csv"
    with open(filename, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["Quote", "Author"])
        writer.writerows(data)

    print(f"Saved {len(data)} records to {filename}")

if __name__ == "__main__":
    main()
```

# Handling Large Datasets (Chunking and Lazy Evaluation)

```python
#question13
import pandas as pd

CSV_FILE = "large_data.csv"   # change to your file
CHUNK_SIZE = 10000

def main():
    for i, chunk in enumerate(pd.read_csv(CSV_FILE, chunksize=CHUNK_SIZE), start=1):
        print(f"Chunk {i} -> rows: {len(chunk)}")
        print(chunk.head(2))
        print("-" * 50)

if __name__ == "__main__":
    main()
```

```python
#question14
import pandas as pd

CSV_FILE = "large_data.csv"
CHUNK_SIZE = 10000
COLUMN_NAME = "amount"   # change to a numeric column in your CSV

def main():
    total = 0.0
    for chunk in pd.read_csv(CSV_FILE, chunksize=CHUNK_SIZE):
        # Convert to numeric safely (handles bad/missing values)
        values = pd.to_numeric(chunk[COLUMN_NAME], errors="coerce").fillna(0)
        total += values.sum()

    print(f"Total sum of '{COLUMN_NAME}' = {total}")

if __name__ == "__main__":
    main()
```

```python
#question16
def square_generator(n):
    for i in range(1, n + 1):
        yield i * i   # computed only when requested

def main():
    gen = square_generator(10)

    print("Generator created (no computation done yet).")
    print("Next values (computed lazily):")
    print(next(gen))
    print(next(gen))
    print(next(gen))

    print("Remaining values:")
    for x in gen:
        print(x)

if __name__ == "__main__":
    main()
```

```python
#question17
TEXT_FILE = "large_text.txt"  # change to your file

def read_lines_lazy(filepath):
    with open(filepath, "r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            yield line.rstrip("\n")

def main():
    count = 0
    for line in read_lines_lazy(TEXT_FILE):
        count += 1
        if count <= 5:  # show first 5 lines
            print(line)

    print("\nTotal lines =", count)

if __name__ == "__main__":
    main()
```

```python
#question18
import time

TEXT_FILE = "large_text.txt"

def normal_read(filepath):
    with open(filepath, "r", encoding="utf-8", errors="ignore") as f:
        lines = f.readlines()   # loads everything into memory
    return len(lines)

def lazy_read(filepath):
    count = 0
    with open(filepath, "r", encoding="utf-8", errors="ignore") as f:
        for _ in f:            # reads line by line (lazy)
            count += 1
    return count

def main():
    start = time.time()
    n1 = normal_read(TEXT_FILE)
    t1 = time.time() - start

    start = time.time()
    n2 = lazy_read(TEXT_FILE)
    t2 = time.time() - start

    print(f"Normal read: lines={n1}, time={t1:.4f} sec (high memory)")
    print(f"Lazy read:   lines={n2}, time={t2:.4f} sec (low memory)")

if __name__ == "__main__":
    main()
```
