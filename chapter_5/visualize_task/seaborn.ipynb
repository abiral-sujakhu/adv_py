```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Q1. Box Plot
Write a Python program using Seaborn to draw a box plot showing the distribution of marks obtained by students in three subjects: Mathematics, Physics, and Chemistry.

```python
# Q1: Box Plot - Distribution of marks in three subjects
mathematics = [65, 70, 75, 80, 85, 90]
physics = [60, 68, 72, 78, 82, 88]
chemistry = [62, 67, 74, 79, 84, 89]

# Create DataFrame
data = {
    'Subject': ['Mathematics']*6 + ['Physics']*6 + ['Chemistry']*6,
    'Marks': mathematics + physics + chemistry
}
df = pd.DataFrame(data)

plt.figure(figsize=(8, 6))
sns.boxplot(x='Subject', y='Marks', data=df, palette='Set2')
plt.title('Distribution of Marks in Three Subjects')
plt.xlabel('Subject')
plt.ylabel('Marks')
plt.show()
```

## Q2. Pair Plot
Write a Python program using Seaborn to create a pair plot for the following dataset to analyze relationships between variables.

```python
# Q2: Pair Plot - Analyze relationships between variables
data = {
    'Height': [150, 155, 160, 165, 170],
    'Weight': [50, 55, 60, 65, 70],
    'Age': [18, 19, 20, 21, 22]
}
df = pd.DataFrame(data)

sns.pairplot(df, diag_kind='kde', corner=False)
plt.suptitle('Pair Plot - Relationships between Height, Weight, and Age', y=1.02)
plt.show()
```

## Q3. Heat Map
Write a Python program using Seaborn to plot a heat map representing the correlation matrix of a dataset containing Sales, Profit, Discount, and Quantity.

```python
# Q3: Heat Map - Correlation matrix
np.random.seed(42)
data = {
    'Sales': np.random.randint(100, 1000, 50),
    'Profit': np.random.randint(10, 200, 50),
    'Discount': np.random.uniform(0, 0.5, 50),
    'Quantity': np.random.randint(1, 50, 50)
}
df = pd.DataFrame(data)

# Calculate correlation matrix
correlation_matrix = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            linewidths=0.5, fmt='.2f', square=True)
plt.title('Correlation Matrix - Sales, Profit, Discount, Quantity')
plt.show()
```
