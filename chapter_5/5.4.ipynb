# 5.4 Interactive Visualization using Plotly

## What is Plotly?

**Plotly** is a powerful Python library for creating **interactive, web-based visualizations**. Unlike static libraries (Matplotlib, Seaborn), Plotly graphs allow users to interact with the data.

## Key Features of Plotly:

| Feature | Description |
|---------|-------------|
| **Hover** | See data values when hovering over points |
| **Zoom** | Zoom in/out on specific regions |
| **Pan** | Move around the visualization |
| **Toggle** | Show/hide data series by clicking legend |
| **Export** | Download as PNG image |
| **3D Plots** | Native support for 3D visualizations |

## Plotly APIs:

1. **plotly.express (px)** - High-level, simple syntax (recommended for beginners)
2. **plotly.graph_objects (go)** - Low-level, more control and customization

## Why Use Plotly?

- **Interactive dashboards** for data exploration
- **Web applications** - renders in browsers
- **Presentations** - engaging visualizations
- **Large datasets** - efficient rendering with WebGL

```python
# Install plotly (run once if not installed)
# !pip install plotly

# Import libraries
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np
```

## 1. Interactive Line Chart

Line charts show trends over time. With Plotly, you can hover to see exact values.

```python
# Interactive Line Chart - Monthly Sales
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [120, 150, 180, 170, 200, 220]

fig = px.line(x=months, y=sales, 
              title='Monthly Sales (Hover to see values)',
              labels={'x': 'Months', 'y': 'Sales'},
              markers=True)
fig.update_traces(line=dict(width=3, color='blue'), 
                  marker=dict(size=10))
fig.show()
```

## 2. Interactive Bar Chart

Bar charts compare categories. Click on legend items to toggle visibility.

```python
# Interactive Bar Chart - Department Students
departments = ['CSE', 'ECE', 'ME', 'CE']
students = [120, 90, 60, 50]

fig = px.bar(x=departments, y=students,
             title='Students in Each Department',
             labels={'x': 'Department', 'y': 'Number of Students'},
             color=departments,
             text=students)
fig.update_traces(textposition='outside')
fig.show()
```

## 3. Interactive Scatter Plot

Scatter plots show relationships between two variables. Hover shows detailed info.

```python
# Interactive Scatter Plot - Study Hours vs Marks
np.random.seed(42)
df = pd.DataFrame({
    'Study_Hours': np.random.uniform(1, 10, 30),
    'Marks': np.random.uniform(40, 100, 30),
    'Student': [f'Student {i+1}' for i in range(30)]
})

fig = px.scatter(df, x='Study_Hours', y='Marks',
                 hover_name='Student',
                 title='Study Hours vs Marks (Hover for student name)',
                 color='Marks',
                 size='Marks',
                 color_continuous_scale='Viridis')
fig.show()
```

## 4. Interactive Pie Chart

Pie charts show proportions. Use `hole` parameter to create donut charts.

```python
# Interactive Pie Chart - Programming Language Popularity
languages = ['Python', 'Java', 'C++', 'JavaScript', 'Others']
popularity = [35, 25, 20, 15, 5]

fig = px.pie(values=popularity, names=languages,
             title='Programming Language Popularity',
             hole=0.3)  # Creates donut chart
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()
```

## 5. Interactive Histogram

Histograms show data distribution. Hover shows bin ranges and counts.

```python
# Interactive Histogram - Student Marks Distribution
marks = [45, 55, 60, 70, 75, 80, 85, 90, 65, 68, 72, 78, 82, 88, 92, 58, 63, 77]

fig = px.histogram(x=marks, nbins=5,
                   title='Distribution of Student Marks',
                   labels={'x': 'Marks', 'y': 'Frequency'},
                   color_discrete_sequence=['steelblue'])
fig.update_layout(bargap=0.1)
fig.show()
```

## 6. Interactive Box Plot

Box plots show distribution with median, quartiles, and outliers.

```python
# Interactive Box Plot - Marks by Subject
data = {
    'Subject': ['Math']*6 + ['Physics']*6 + ['Chemistry']*6,
    'Marks': [65, 70, 75, 80, 85, 90] + [60, 68, 72, 78, 82, 88] + [62, 67, 74, 79, 84, 89]
}
df = pd.DataFrame(data)

fig = px.box(df, x='Subject', y='Marks',
             title='Distribution of Marks by Subject',
             color='Subject')
fig.show()
```

## 7. Interactive Heatmap

Heatmaps visualize matrices like correlation. Use `text_auto` to show values.

```python
# Interactive Heatmap - Correlation Matrix
np.random.seed(42)
df = pd.DataFrame({
    'Sales': np.random.randint(100, 1000, 50),
    'Profit': np.random.randint(10, 200, 50),
    'Discount': np.random.uniform(0, 0.5, 50),
    'Quantity': np.random.randint(1, 50, 50)
})

corr_matrix = df.corr()

fig = px.imshow(corr_matrix,
                text_auto='.2f',
                title='Correlation Heatmap',
                color_continuous_scale='RdBu_r',
                aspect='auto')
fig.show()
```

## 8. Interactive 3D Scatter Plot

3D plots are a unique strength of Plotly - rotate, zoom, and explore data in 3 dimensions!

```python
# 3D Scatter Plot - Height, Weight, Age
df = pd.DataFrame({
    'Height': [150, 155, 160, 165, 170, 158, 162, 168, 172, 175],
    'Weight': [50, 55, 60, 65, 70, 58, 63, 68, 73, 78],
    'Age': [18, 19, 20, 21, 22, 19, 20, 21, 22, 23]
})

fig = px.scatter_3d(df, x='Height', y='Weight', z='Age',
                    color='Age',
                    size='Weight',
                    title='3D Scatter: Height, Weight, Age (Drag to rotate!)')
fig.show()
```

## 9. Using graph_objects for More Control

`plotly.graph_objects` provides more customization options.

```python
# Using graph_objects - Multiple traces in one plot
fig = go.Figure()

# Add first line
fig.add_trace(go.Scatter(
    x=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    y=[120, 150, 180, 170, 200, 220],
    mode='lines+markers',
    name='2024 Sales',
    line=dict(color='blue', width=3)
))

# Add second line
fig.add_trace(go.Scatter(
    x=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    y=[100, 130, 160, 150, 180, 200],
    mode='lines+markers',
    name='2023 Sales',
    line=dict(color='red', width=3, dash='dash')
))

fig.update_layout(
    title='Sales Comparison: 2023 vs 2024',
    xaxis_title='Month',
    yaxis_title='Sales',
    hovermode='x unified'
)
fig.show()
```

## Summary: Plotly vs Matplotlib/Seaborn

| Feature | Matplotlib/Seaborn | Plotly |
|---------|-------------------|--------|
| **Interactivity** | Static images | Fully interactive (zoom, hover, pan) |
| **3D Plots** | Limited support | Excellent native support |
| **Hover Info** | Not available | Built-in hover tooltips |
| **Web Integration** | Requires export | Native HTML/JS rendering |
| **Learning Curve** | Moderate | Easy with plotly.express |
| **Output Format** | PNG, PDF, SVG | HTML, PNG, interactive widgets |
| **Best Use Case** | Publications, reports | Dashboards, presentations, exploration |

### Key Plotly Functions:
- `px.line()` - Line charts
- `px.bar()` - Bar charts
- `px.scatter()` - Scatter plots
- `px.pie()` - Pie charts
- `px.histogram()` - Histograms
- `px.box()` - Box plots
- `px.imshow()` - Heatmaps
- `px.scatter_3d()` - 3D scatter plots
