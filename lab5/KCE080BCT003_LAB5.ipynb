# Task 1: Descriptive Statistics

Create a dataset of 50 random integers between 10 and 100.
Calculate:
- Mean
- Median
- Standard Deviation

```python
import numpy as np

data = np.random.randint(10, 101, 50)

mean = np.mean(data)
median = np.median(data)
std_dev = np.std(data)

print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std_dev)
```

# Task 2: Covariance and Correlation

Generate two random arrays and compute:
- Covariance
- Pearson correlation
- Spearman correlation

```python
from scipy import stats

x = np.random.rand(100)
y = np.random.rand(100)

covariance = np.cov(x, y)[0,1]
pearson = stats.pearsonr(x, y)
spearman = stats.spearmanr(x, y)

print("Covariance:", covariance)
print("Pearson Correlation:", pearson)
print("Spearman Correlation:", spearman)
```

# Task 3: Skewness

Create a skewed dataset and calculate skewness.

```python
skewed_data = np.random.exponential(scale=2, size=1000)

skewness = stats.skew(skewed_data)

print("Skewness:", skewness)
```

# Task 4: Kurtosis

Generate normal distribution and calculate kurtosis.

```python
normal_data = np.random.normal(0, 1, 1000)

kurtosis = stats.kurtosis(normal_data)

print("Kurtosis:", kurtosis)
```

# Task 9: Sampling Methods
- Random Sampling
- Systematic Sampling
- Stratified Sampling

```python
population = np.random.rand(1000)

# Random Sample
random_sample = np.random.choice(population, 100, replace=False)

# Systematic Sample
systematic_sample = population[::10][:100]

# Stratified Sample
bins = np.array_split(population, 10)
stratified_sample = np.concatenate([np.random.choice(b, 10) for b in bins])

print("Random Sample Size:", len(random_sample))
print("Systematic Sample Size:", len(systematic_sample))
print("Stratified Sample Size:", len(stratified_sample))
```

# Task 10: Central Limit Theorem
Take 50 samples and plot sampling distribution of mean

```python
means = []

for _ in range(50):
    sample = np.random.choice(population, 100)
    means.append(np.mean(sample))

plt.hist(means, bins=15)
plt.title("Sampling Distribution of Mean")
plt.show()
```

# Task 11: One-Sample t-test

```python
sample = np.random.normal(50, 5, 30)
t_stat, p_value = stats.ttest_1samp(sample, 50)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

# Task 12: Two-Sample t-test

```python
group1 = np.random.normal(50, 5, 30)
group2 = np.random.normal(52, 5, 30)

t_stat, p_value = stats.ttest_ind(group1, group2)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

# Task 13: Chi-Square Test

```python
import pandas as pd

data = [[10, 20], [20, 40]]
chi2, p, dof, expected = stats.chi2_contingency(data)

print("Chi-square:", chi2)
print("P-value:", p)
```

# Task 14: 95% Confidence Interval

```python
sample = np.random.normal(50, 5, 30)
mean = np.mean(sample)
sem = stats.sem(sample)

ci = stats.t.interval(0.95, len(sample)-1, loc=mean, scale=sem)

print("95% Confidence Interval:", ci)
```

# Task 15: Simple Linear Regression

```python
import statsmodels.api as sm

x = np.random.rand(100)
y = 2*x + np.random.rand(100)

x = sm.add_constant(x)
model = sm.OLS(y, x).fit()

print(model.summary())
```

# Task 17: Multiple Linear Regression

```python
x1 = np.random.rand(100)
x2 = np.random.rand(100)
y = 3*x1 + 2*x2 + np.random.rand(100)

X = np.column_stack((x1, x2))
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()
print(model.summary())
```

# Task 18: Time Series Trend Analysis

```python
time = np.arange(100)
values = 0.5*time + np.random.normal(0,5,100)

X = sm.add_constant(time)
model = sm.OLS(values, X).fit()

plt.plot(time, values)
plt.plot(time, model.predict(X))
plt.show()
```
