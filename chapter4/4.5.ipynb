# Automation of EDA workflows using Python 

## ydata profiling

```python
import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv("customer_shopping_data.csv")
profile = ProfileReport(df, 
                        title="Customer Shopping Data EDA Report",
                        explorative=True)

profile.to_file("customer_shopping_eda_report.html")

```

## Sweetviz

```python
import pandas as pd
import sweetviz as sv

df = pd.read_csv("customer_shopping_data.csv")

report = sv.analyze(df)
report.show_html("sweetviz_customer_shopping_report.html", open_browser=False)
```

## AutoViz

```python
from autoviz.AutoViz_Class import AutoViz_Class

AV = AutoViz_Class()

# It will generate many plots automatically
df_viz = AV.AutoViz(
    filename="/customer_shopping_data.csv",
    sep=",",
    depVar="",          # leave empty if you don't have a target variable
    dfte=None,
    header=0,
    verbose=1
)
```
