# Population and sampling

```python
import pandas as pd
import numpy as np

np.random.seed(42)

N = 50000  # large population size

regions = ["Asia", "Europe", "Africa", "North America", "South America", "Oceania"]
countries = {
    "Asia": ["Nepal", "India", "China", "Japan", "Bangladesh"],
    "Europe": ["Germany", "France", "UK", "Italy", "Spain"],
    "Africa": ["Nigeria", "Kenya", "Egypt", "Ethiopia", "South Africa"],
    "North America": ["USA", "Canada", "Mexico", "Cuba", "Panama"],
    "South America": ["Brazil", "Argentina", "Chile", "Peru", "Colombia"],
    "Oceania": ["Australia", "New Zealand", "Fiji", "Samoa", "Tonga"]
}

age_groups = ["0-14", "15-24", "25-54", "55-64", "65+"]
genders = ["Male", "Female"]
areas = ["Urban", "Rural"]

# Generate columns
region_col = np.random.choice(regions, size=N, p=[0.45, 0.12, 0.18, 0.12, 0.10, 0.03])

country_col = [np.random.choice(countries[r]) for r in region_col]

df = pd.DataFrame({
    "PersonID": np.arange(1, N+1),
    "Region": region_col,
    "Country": country_col,
    "Gender": np.random.choice(genders, size=N, p=[0.51, 0.49]),
    "AgeGroup": np.random.choice(age_groups, size=N, p=[0.25, 0.15, 0.40, 0.10, 0.10]),
    "Area": np.random.choice(areas, size=N, p=[0.65, 0.35]),
})

# Add a realistic "Income" and "HouseholdSize"
df["HouseholdSize"] = np.random.randint(1, 8, size=N)
df["IncomeUSD"] = (np.random.normal(loc=12000, scale=6000, size=N)).clip(200, 100000).round(0).astype(int)

df.head()
```

## random sampling

```python
random_sample_n = df.sample(n=1000, random_state=42)
random_sample_n.head(), len(random_sample_n)
```

## stratified sampling 

```python
stratified_fixed = df.groupby("Region", group_keys=False).apply(
    lambda x: x.sample(n=200, random_state=42)
)

stratified_fixed.head(), len(stratified_fixed)
```

## systematic sampling 

```python
sample_size = 1000
k = len(df) // sample_size   # step size
k
```

```python
start = np.random.randint(0, k)
systematic_sample_random_start = df.iloc[start::k].head(sample_size)

systematic_sample_random_start.head(), len(systematic_sample_random_start)
```

# Hypothesis testing 


# 📊 Hypothesis Testing Using One-Sample Z-Test

## 1️⃣ Objective
To test whether the population mean income is equal to the hypothesized value using a **One-Sample Z-Test**.

---

## 2️⃣ Problem Statement
A large population dataset of income values was generated, and a random sample of size **n = 1000** was selected.

We want to test whether the population mean income is equal to:

\[
\mu = 12000
\]

---

## 3️⃣ Hypotheses

### Null Hypothesis (H₀):
\[
H_0 : \mu = 12000
\]

The population mean income is equal to 12000.

### Alternative Hypothesis (H₁):
\[
H_1 : \mu \ne 12000
\]

The population mean income is not equal to 12000.

This is a **Two-Tailed Test**.

---

## 4️⃣ Given Information

- Population Standard Deviation (σ) = 6000  
- Sample Size (n) = 1000  
- Significance Level (α) = 0.05  
- Sample Mean (x̄) = Calculated from the sample  

---

## 5️⃣ Z-Test Formula

\[
Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}
\]

Where:

- \( \bar{x} \) = Sample Mean  
- \( \mu_0 \) = Hypothesized Population Mean  
- \( \sigma \) = Population Standard Deviation  
- \( n \) = Sample Size  

---

## 6️⃣ Decision Rule

At 5% level of significance (α = 0.05):

- If **p-value < 0.05** → Reject H₀  
- If **p-value ≥ 0.05** → Fail to Reject H₀  

OR

- If **|Z| > 1.96** → Reject H₀ (for two-tailed test)

---

## 7️⃣ Interpretation

- If H₀ is rejected → There is significant statistical evidence that the population mean income differs from 12000.
- If H₀ is not rejected → There is no significant statistical evidence to conclude that the population mean income differs from 12000.

---

## 8️⃣ Conclusion

Using the Z-test formula, the Z-statistic and corresponding p-value were calculated.  
Based on the comparison with the significance level (α = 0.05), a statistical decision was made regarding the null hypothesis.

Thus, hypothesis testing helps determine whether the observed sample data provides enough evidence to support or reject the assumed population mean.


```python
# -------------------------------
# Hypothesis Testing using Z-Test
# -------------------------------

import numpy as np
from scipy.stats import norm

# Step 1: Generate Large Population Data
np.random.seed(42)
population = np.random.normal(loc=12000, scale=6000, size=50000)

# Step 2: Take Random Sample
sample = np.random.choice(population, size=1000, replace=False)

# Step 3: Define Known Values
mu_0 = 12000               # Hypothesized population mean
sigma = 6000               # Known population standard deviation
n = len(sample)            # Sample size
x_bar = np.mean(sample)    # Sample mean
alpha = 0.05               # Significance level

# Step 4: Calculate Z-Statistic
z = (x_bar - mu_0) / (sigma / np.sqrt(n))

# Step 5: Calculate p-value (Two-Tailed Test)
p_value = 2 * (1 - norm.cdf(abs(z)))

# Step 6: Decision
if p_value < alpha:
    decision = "Reject Null Hypothesis (H0)"
else:
    decision = "Fail to Reject Null Hypothesis (H0)"

# Step 7: Print Results
print("-------- Z-Test Hypothesis Testing --------")
print(f"Null Hypothesis (H0): μ = {mu_0}")
print(f"Alternative Hypothesis (H1): μ ≠ {mu_0}")
print(f"Sample Mean (x̄): {round(x_bar,2)}")
print(f"Population Std Dev (σ): {sigma}")
print(f"Sample Size (n): {n}")
print(f"Z-Statistic: {round(z,4)}")
print(f"P-Value: {round(p_value,6)}")
print("Decision at α = 0.05 :", decision)
```
